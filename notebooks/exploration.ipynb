```python
import pandas as pd

df = pd.read_csv("../data/recruitment_pipeline_biased.csv")
df.head()
```

```python
df['StageReached'].value_counts()
```

```python
stage_counts = df['StageReached'].value_counts().reindex([
    'Applied', 'Screened', 'Interviewed', 'Shortlisted', 'Hired'
])

dropoff = stage_counts.pct_change().abs() * 100

dropoff
```

```python
conversion = stage_counts / stage_counts.shift(1) * 100
conversion
```

```python
df.columns
```

```python
gender_selection = df.groupby('Gender')['StageReached'].apply(lambda x: (x == 'Hired').mean() * 100)
gender_selection
```

```python
gender_rates = df.groupby('Gender')['StageReached'].apply(lambda x: (x == 'Hired').mean())
gender_air = gender_rates / gender_rates.max()
gender_air
```

```python
ethnicity_selection = df.groupby('Ethnicity')['StageReached'].apply(lambda x: (x == 'Hired').mean() * 100)
ethnicity_selection
```

```python
ethnicity_rates = df.groupby('Ethnicity')['StageReached'].apply(lambda x: (x == 'Hired').mean())
ethnicity_air = ethnicity_rates / ethnicity_rates.max()
ethnicity_air
```

```python
disability_selection = df.groupby('DisabilityStatus')['StageReached'].apply(lambda x: (x == 'Hired').mean() * 100)
disability_selection
```

```python
disability_rates = df.groupby('DisabilityStatus')['StageReached'].apply(lambda x: (x == 'Hired').mean())
disability_air = disability_rates / disability_rates.max()
disability_air
```

```python
role_selection = df.groupby('RoleType')['StageReached'].apply(lambda x: (x == 'Hired').mean() * 100)
role_selection
```

```python
stage_counts_table = df['StageReached'].value_counts().reset_index()
stage_counts_table.columns = ['StageReached', 'Count']
stage_counts_table
```

```python
stage_order = ['Applied', 'Screened', 'Interviewed', 'Shortlisted', 'Hired']

stage_counts = df['StageReached'].value_counts().reindex(stage_order)
dropoff_table = stage_counts.pct_change().abs() * 100
dropoff_table = dropoff_table.reset_index()
dropoff_table.columns = ['StageReached', 'DropoffPercent']
dropoff_table
```

```python
conversion_table = (stage_counts / stage_counts.shift(1) * 100).reset_index()
conversion_table.columns = ['StageReached', 'ConversionPercent']
conversion_table
```

```python
fairness_summary = {
    'Gender_AIR': gender_air,
    'Ethnicity_AIR': ethnicity_air,
    'Disability_AIR': disability_air
}

fairness_df = pd.concat(fairness_summary, axis=1)
fairness_df
```

```python
hires_by_demo = df[df['StageReached'] == 'Hired'].groupby(
    ['Gender', 'Ethnicity', 'DisabilityStatus']
).size().reset_index(name='Hires')
hires_by_demo
```

```python
stage_counts_table.to_csv("stage_counts.csv", index=False)
dropoff_table.to_csv("dropoff.csv", index=False)
conversion_table.to_csv("conversion.csv", index=False)
fairness_df.to_csv("fairness_summary.csv")
hires_by_demo.to_csv("hires_by_demo.csv", index=False)
```
